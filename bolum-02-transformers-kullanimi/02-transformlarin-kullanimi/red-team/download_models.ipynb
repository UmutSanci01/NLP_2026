```python
import os
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# İndirilecek model haritası
MODELS = {
    "DistilBERT (WordPiece)": "distilbert-base-uncased-finetuned-sst-2-english",
    "RoBERTa (BPE)": "textattack/roberta-base-SST-2",
    "ALBERT (SentencePiece)": "textattack/albert-base-v2-SST-2"
}

BASE_DIR = "./models"

for name, hf_path in MODELS.items():
    save_path = os.path.join(BASE_DIR, name)
    if not os.path.exists(save_path):
        print(f"[{name}] modeli indiriliyor ve kaydediliyor...")
        tokenizer = AutoTokenizer.from_pretrained(hf_path)
        model = AutoModelForSequenceClassification.from_pretrained(hf_path)
        
        tokenizer.save_pretrained(save_path)
        model.save_pretrained(save_path)
        print(f"[{name}] başarıyla '{save_path}' dizinine kaydedildi.\n")
    else:
        print(f"[{name}] zaten yerelde mevcut: {save_path}")
```
