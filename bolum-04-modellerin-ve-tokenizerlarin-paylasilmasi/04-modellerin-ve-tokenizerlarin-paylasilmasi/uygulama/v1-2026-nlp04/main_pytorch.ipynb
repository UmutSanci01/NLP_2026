## 2. Önceden Eğitilmiş Modellerin Kullanımı (Section 2: Using pretrained models)

### Kod 1: `pipeline` ile Maske Doldurma (`fill-mask`)

```python
from transformers import pipeline

camembert_fill_mask = pipeline("fill-mask", model="camembert-base")
results = camembert_fill_mask("Le camembert est <mask> :)")
results
```

### Kod 2: Modele Özel Mimari Sınıfları ile Yükleme
CamemBERT mimarisine özel sınıflar (`CamembertTokenizer`, `CamembertForMaskedLM`):

```python
from transformers import CamembertTokenizer, CamembertForMaskedLM

tokenizer = CamembertTokenizer.from_pretrained("camembert-base")
model = CamembertForMaskedLM.from_pretrained("camembert-base")
print("[OK] Modele ozel mimari siniflari basariyla yuklendi!")
```

### Kod 3: Mimariden Bağımsız `Auto*` Sınıfları ile Yükleme
Herhangi bir model kontrol noktasıyla çalışan dinamik `AutoTokenizer` ve `AutoModelForMaskedLM`:

```python
from transformers import AutoTokenizer, AutoModelForMaskedLM

tokenizer = AutoTokenizer.from_pretrained("camembert-base")
model = AutoModelForMaskedLM.from_pretrained("camembert-base")
print("[OK] Auto* siniflari basariyla yuklendi!")
```
